```python
from datasets import load_dataset
ds = load_dataset("mteb/amazon_massive_intent", "fr")
```

```python
ds
```

```python
import pandas as pd
df_train =  pd.DataFrame( ds["train"])
df_test =  pd.DataFrame( ds["test"])
df_validation =  pd.DataFrame( ds["validation"])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report , accuracy_score , f1_score
```

```python
pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2))),
    ("clf", LogisticRegression(max_iter=1000)),
])
```

```python
pipeline.fit(df_train["text"], df_train["label_text"])
```

```python
pred = pipeline.predict(df_validation["text"])
print("Accuracy :", accuracy_score(df_validation["label_text"], pred))
print("F1 macro :", f1_score(df_validation["label_text"], pred, average="macro"))
```

```python
import pickle
```

```python
with open("model.pkl" , "wb") as f :
    pickle.dump(pipeline , f)
```
